Top-1 accuracy is the fraction of images where the model’s single most probable class is the correct label.

Top-5 accuracy is the fraction where the correct label is anywhere among the model’s five most probable classes.

Example: an image’s true label is “Siberian husky”, and the model’s probabilities rank “Alaskan malamute” first, then “Siberian husky”, “Eskimo dog”, and so on.

           Top-1: wrong, because the first guess isn’t husky.
           Top-5: correct, because husky is in the top five.

```python
import torch

def top_k_accuracy_error(labels, logits, k = 1):
    top_k = logits.top_k(k, dim= 1).indices
    correct = (top_k == labels[:, None]).any(dim=1)
    return 1 - correct.float().mean().item()
```
